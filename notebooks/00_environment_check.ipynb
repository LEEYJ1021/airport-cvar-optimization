# Notebook 0: Environment Check

This notebook verifies that the Python environment is correctly set up, all required packages are installed, and the database connection is working.

**Steps:**
1. Check Python version.
2. Import all major dependencies to ensure they are installed.
3. Test the connection to the MySQL database using the credentials from the `.env` file.

### 1. Python and Package Imports

```python
import sys
import os
import logging

import pandas as pd
import numpy as np
import sqlalchemy
import typer
import requests
from dotenv import load_dotenv

# Check Python version
print(f"Python Version: {sys.version}")
assert sys.version_info >= (3, 10), "Python 3.10+ is required."

print(f"Pandas Version: {pd.__version__}")
print(f"SQLAlchemy Version: {sqlalchemy.__version__}")

print("\n✅ All major packages imported successfully.")
```

### 2. Environment Variable and Configuration Check

```python
# We need to go up two directories to find the .env file from notebooks/
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.append(project_root)

from src.settings import get_config, get_service_key

try:
    config = get_config()
    print("Configuration loaded successfully:")
    for key, value in config.items():
        # Hide sensitive values
        if 'password' in key or 'key' in key:
            print(f"  - {key}: {'*' * 8}")
        else:
            print(f"  - {key}: {value}")

    # Test API key retrieval
    assert get_service_key('default') is not None, "Default service key not found!"
    assert get_service_key('kma') is not None, "KMA service key not found!"
    print("\n✅ API keys are configured.")
except Exception as e:
    print(f"❌ Error loading configuration: {e}")
    print("Please ensure your .env file is correctly set up in the project root.")
```

### 3. Database Connection Test

```python
from src.db import get_engine

print("Attempting to connect to the database...")

try:
    engine = get_engine()
    with engine.connect() as connection:
        result = connection.execute(sqlalchemy.text("SELECT 1"))
        if result.scalar() == 1:
            print("✅ Database connection successful!")
        else:
            print("❌ Connection established, but query failed.")
except Exception as e:
    print(f"❌ Database connection failed: {e}")
    print("\nPlease check your MYSQL_* environment variables in the .env file.")
```
